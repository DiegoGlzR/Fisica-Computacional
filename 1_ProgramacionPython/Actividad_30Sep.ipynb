### Función de Fey en el plano XY

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 24*np.pi, 20000)

r = np.exp(np.cos(theta)) - 2*np.cos(4*theta) + np.sin(theta/12)**5

x = r * np.cos(theta)
y = r * np.sin(theta)

plt.figure(figsize=(9, 9))

plt.plot(x, y, linewidth=1.5, color='green', label= r'$r=e^{\cos\theta}-2\cos(4\theta)+\sin^5\left(\frac{\theta}{12}\right)$')

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.grid(True, linestyle='--', alpha=0.5)

plt.xlabel(r'$x$', fontsize=14)
plt.ylabel(r'$y$', fontsize=14)

# Título
plt.title(
    'Función de Fey',
    fontsize=16
)
plt.legend(loc=2)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Constantes
k = 8.99e9
q = 1

x = np.linspace(-0.5, 0.5, 500)
y = np.linspace(-0.5, 0.5, 500)

X, Y = np.meshgrid(x, y)

r = np.sqrt(X**2 + Y**2)

# Potencial eléctrico
V = k*q/r

plt.imshow(
    V,
    extent=[-0.5, 0.5, -0.5, 0.5],
    origin='lower',
    cmap='hot'
)

plt.colorbar(label='Potencial [V]')

plt.xlabel('x [m]')
plt.ylabel('y [m]')
plt.title('Potencial electrostático')

plt.show()
```
