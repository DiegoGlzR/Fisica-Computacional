### La constante de Madelung
En física de la materia condensada, la constante de Madelung da el potencial electrico total que siente un atomo en un s  ́ olido; y depende de las cargas de los otros ́atomos cercanos y de sus ubicaciones. Por ejemplo, el cristal de cloruro de sodio solido (la sal de mesa), tiene  ́
atomos dispuestos en una red cúbica, con átomos de sodio y cloro alternados, teniendo los de sodio una carga positiva +e y los de cloro una negativa −e, (donde e es la carga del electron). Si etiquetamos cada posición en la red con tres coordenadas enteras (i, j, k),
entonces los atomos de sodio caen en posiciones donde i + j + k es par, y los atomos de cloro en posiciones donde i + j + k es impar.
Consideremos un atomo de sodio en el origen, i.e. i = j = k = 0, y calculemos la constante de Madelung. Si el espaciado de los atomos en la red es a, entonces la distancia desde el origen al atomo en la posición (i, j, k) es:

$\sqrt{(ia)^{2} + (ja)^{2} + (ka)^{2}} = a \sqrt{(i)^{2} + (j)^{2} + (k)^{2}}$

y el potencial en el origen creado por el átomo es:

$V(i,j,k) = \pm \frac{e}{4 \pi \epsilon_{0}a \sqrt{i^{2} + j^{2} + k^{2}}}$

siendo $\epsilon_{0}$ la permitividad del vacío y el signo de la expresión se toma dependiento de si i+j+k es par o impar. Así entonces, el potencial que siente el átomo de sodio es la suma de esta cantidad sobre todos los demás átomos. Supongamos una caja cúbica alrededor del átomo de sodio en el origen, con L átomos en todas la direcciones, entonces:
\begin{equation}
    V_{total} = \sum_{i,j,k=-L}^{L} V(i,j,k) = \frac{e}{4\pi \epsilon a} M
\end{equation}

donde M es la constante de Madelung (al menos aproximadamente).
Entonces la constante de Madelung M sería:
\begin{equation}
    M = \frac{4\pi \epsilon a \sum_{i,j,k=-L}^{L} (-1)^{i+k+j} \frac{e}{4 \pi \epsilon_{0}a \sqrt{i^{2} + j^{2} + k^{2}}}}{e}
\end{equation}

```python
import math

def calcular_madelung(L):
    M = 0.0
    # Recorremos la caja tridimensional desde -L hasta L
    for i in range(-L, L + 1):
        for j in range(-L, L + 1):
            for k in range(-L, L + 1):
                # Omitimos el átomo en el origen (0, 0, 0)
                if i == 0 and j == 0 and k == 0:
                    continue
                
                distancia = math.sqrt(i**2 + j**2 + k**2)
                
                # Cargas iguales repelen (-), cargas opuestas atraen (+)
                if (i + j + k) % 2 == 0:
                    M -= 1.0 / distancia
                else:
                    M += 1.0 / distancia
                    
    return M

# Definimos L y ejecutamos
L = 100
M_aprox = calcular_madelung(L)

print(f"Constante de Madelung aproximada (L={L}): {M_aprox:.5f}")
```

### 2. La fórmula semiempírica de la masa (FSM)

En física nuclear, la fórmula de Weizsäcker (conocida también como fórmula semiempírica) sirve para evaluar la masa y otras propiedades de un núcleo atómico; y está basada parcialmente en mediciones empíricas. En particular la fórmula se usa para calcular la **energía de enlace nuclear aproximada $B$**, de un núcleo atómico con número atómico $Z$ y número de masa $A$:

$$B = a_1 A - a_2 A^{2/3} - a_3 \frac{Z^2}{A^{1/3}} - a_4 \frac{(A - 2Z)^2}{A} + \frac{a_5}{A^{1/2}}$$

donde, en unidades de millones de electrón-volts (MeV), las constantes son $a_1 = 15.8$, $a_2 = 18.3$, $a_3 = 0.714$, $a_4 = 23.2$ y:

$$a_5 = \begin{cases} 0 & \text{si } A \text{ es impar,} \\ 12.0 & \text{si } A \text{ y } Z \text{ son pares (ambos),} \\ -12.0 & \text{si } A \text{ es par y } Z \text{ es impar.} \end{cases}$$


* **a)** Escribe un programa que tome como entrada los valores de $A$ y $Z$, e imprima la energía de enlace $B$ para el átomo correspondiente. Usa tu programa para encontrar la energía de enlace de un átomo con $A = 58$ y $Z = 28$ *(Hint: La respuesta correcta es alrededor de los 490 MeV)*.

* **b)** Modifica el programa del inciso anterior, para escribir una segunda versión que imprima no la energía de enlace total $B$, sino la energía de unión por nucleón, que es $B / A$.

* **c)** Escribe una tercera versión del programa para que tome como entrada solo un valor del número atómico $Z$ y luego pase por todos los valores de $A$ desde $A = Z$ hasta $A = 3Z$, para encontrar el que tiene la mayor energía de enlace por nucleón. Este es el núcleo más estable con el número atómico dado. Haz que tu programa imprima el valor de $A$ para este núcleo más estable y el valor de la energía de enlace por nucleón.

* **d)** Finalmente, escribe una cuarta versión del programa que, en lugar de tomar $Z$ como entrada, se ejecute a través de todos los valores de $Z$ de 1 a 100 e imprima el valor más estable de $A$ para cada uno. ¿A qué valor de $Z$ se produce la energía de enlace máxima por nucleón? *(La respuesta correcta, en la vida real, es $Z = 28$, que corresponde al Níquel)*.

```python
#INCISO A
def energia_enlace(A, Z):
    # Definimos las constantes en MeV
    a1 = 15.8
    a2 = 18.3
    a3 = 0.714
    a4 = 23.2
    
    # Determinamos a5 según las condiciones de paridad de A y Z
    if A % 2 != 0:
        a5 = 0.0
    elif A % 2 == 0 and Z % 2 == 0:
        a5 = 12.0
    else:  # A es par y Z es impar
        a5 = -12.0
        
    B = a1 * A - a2 * (A**(2/3)) - a3 * (Z**2) / (A**(1/3)) - a4 * ((A - 2*Z)**2) / A + a5 / (A**(1/2))
    
    return B

A = 58
Z = 28

B = energia_enlace(A, Z)
print(f"La energía de enlace B para A={A} y Z={Z} es: {B:.2f} MeV")
```

```python
# INCISO B
def energia_por_nucleon(A, Z):
    B = energia_enlace(A, Z)
    return B / A

A = 58
Z = 28

b = energia_por_nucleon(A, Z)
print(f"La energía de enlace por nucleón (B/A) para A={A} y Z={Z} es: {b:.4f} MeV")
```

```python
# INCISO C
def nucleo_mas_estable(Z):
    max_B_por_A = -float('inf')
    mejor_A = Z
    
    # Recorremos A desde Z hasta 3Z (incluyendo 3Z)
    for A in range(Z, 3*Z + 1):
        actual_B_por_A = energia_por_nucleon(A, Z)
        
        # Si la energía por nucleón es mayor, actualizamos el valor máximo
        if actual_B_por_A > max_B_por_A:
            max_B_por_A = actual_B_por_A
            mejor_A = A
            
    return mejor_A, max_B_por_A

# Probamos con Z = 28
Z = 28
a_estable, max_ba = nucleo_mas_estable(Z)

print(f"Para Z = {Z}:")
print(f"El número de masa A más estable es: {a_estable}")
print(f"La energía de enlace por nucleón máxima es: {max_ba:.4f} MeV")
```

```python
#INCISO D
max_global_B_por_A = -float('inf')
mejor_Z_global = 0
mejor_A_global = 0

print("Z \t A más estable \t B/A máx (MeV)")
print("-" * 38)

# Recorremos Z de 1 a 100
for Z in range(1, 101):
    A_estable, max_ba = nucleo_mas_estable(Z)
    
    print(f"{Z} \t {A_estable} \t\t {max_ba:.4f}")
    
    # Comprobamos si es el máximo global
    if max_ba > max_global_B_por_A:
        max_global_B_por_A = max_ba
        mejor_Z_global = Z
        mejor_A_global = A_estable

print("-" * 38)
print(f"La energía de enlace máxima por nucleón ocurre en Z = {mejor_Z_global} (A = {mejor_A_global})")
print(f"Valor máximo de B/A = {max_global_B_por_A:.4f} MeV")
```

### 3. Coeficientes binomiales

El coeficiente binomial $\binom{n}{k}$ es un número entero igual a:

$$\binom{n}{k} = \frac{n!}{k!(n-k)!} = \frac{n \times (n-1) \times (n-2) \times \dots \times (n-k+1)}{1 \times 2 \times \dots \times k}$$

donde $k \ge 1$, o bien $\binom{n}{0} = 1$ cuando $k = 0$.


* **a)** Utiliza esta fórmula para escribir una función llamada `binomial(n,k)` (o como tú quieras) que calcule el coeficiente binomial para un $n$ y $k$ dados. Asegúrate de que tu función devuelva la respuesta en forma de un número entero (no flotante) y proporcione el valor correcto de 1 para el caso en que $k = 0$.

* **b)** Usando tu función, escribe un programa que imprima las primeras 20 líneas del *"triángulo de Pascal"*. La $n$-ésima línea del triángulo de Pascal contiene $n+1$ números, que son los coeficientes $\binom{n}{0}$, $\binom{n}{1}$, y así sucesivamente hasta $\binom{n}{n}$. De tal manera que las primeras líneas son:

11

* **c)** La probabilidad de que para una moneda no sesgada, lanzada $n$ veces, salga águila $k$ veces es:
$$p(k|n) = \frac{\binom{n}{k}}{2^n}$$

Escribe un programa para calcular:
1. La probabilidad total de que una moneda lanzada 100 veces, salga águila exactamente 60 veces y
2. La probabilidad de que salga águila 60 veces o más.

```python
# INCISO A
def binomial(n, k):
    # Caso base
    if k == 0:
        return 1
    
    numerador = 1
    denominador = 1
    
    for i in range(1, k + 1):
        numerador *= (n - i + 1)
        denominador *= i
        
    # Usamos // para asegurar que devuelva un número entero (int) y no flotante
    return numerador // denominador

print("binomial(5, 0) =", binomial(5, 0))
```

```python
# INCISO B
# Imprimir las primeras 20 líneas del triángulo de Pascal (de n = 0 a n = 19)
for n in range(20):
    linea = []
    for k in range(n + 1):
        linea.append(str(binomial(n, k)))
    
    # Imprimimos los números de la línea separados por un espacio
    print(" ".join(linea))
```

```python
# INCISO C
# Usamos la función 'binomial' definida en el inciso a
n = 100

# 1) Probabilidad de obtener exactamente 60 águilas en 100 lanzamientos: p(60|100)
k1 = 60
prob_60 = binomial(n, k1) / (2**n)

# 2) Probabilidad de obtener 60 o más águilas: suma de p(k|100) para k desde 60 hasta 100
prob_60_o_mas = 0.0
for k in range(60, n + 1):
    prob_60_o_mas += binomial(n, k) / (2**n)

# Resultados
print(f"1) Probabilidad de obtener exactamente 60 águilas: {prob_60:.6f} ({prob_60 * 100:.4f}%)")
print(f"2) Probabilidad de obtener 60 o más águilas: {prob_60_o_mas:.6f} ({prob_60_o_mas * 100:.4f}%)")
```

### 4. Números primos

Es posible escribir un programa bastante rápido para calcular números primos siguiendo las siguientes observaciones:

* **a)** Un número $n$ es primo, si no tiene factores primos menores que $n$. Por lo tanto, sólo necesitamos comprobar si es divisible por otros números primos.

* **b)** Si un número $n$ no es primo y tiene un factor $r$, entonces $n = rs$, donde $s$ también es un factor. Si $r \ge \sqrt{n}$ entonces $n = rs \ge \sqrt{n}s$, lo que implica que $s \le \sqrt{n}$. Es decir, cualquier **no primo** debe tener factores, y por lo tanto también factores primos, menores o iguales a $\sqrt{n}$. Por lo tanto, para determinar si un número es primo, tenemos que verificar sus factores primos solo hasta $\sqrt{n}$; incluso si no hay ninguno, entonces el número es primo.

* **c)** Si encontramos un **factor primo menor que $\sqrt{n}$** entonces sabemos que el **número no es primo** y, por lo tanto, no hay necesidad de realizar más comprobaciones; podemos abandonar este número y pasar al siguiente.

Escribe un programa que encuentre todos los **números primos hasta 10 000 (diez mil)**. Crea una lista para almacenar los números primos, que comience con solo un número primo (el 2). Luego, para cada número $n$ del 3 al 10 000, verifica si el número es divisible por alguno de los números primos de la lista hasta $\sqrt{n}$. En cuanto encuentres un factor primo, puedes dejar de comprobar el resto de números (pues ya sabrías que $n$ no es primo). Si no encuentras factores primos iguales o menores a $\sqrt{n}$, entonces $n$ es primo y habría que agregarlo a la lista. Puedes imprimir la lista completa al final del programa, o bien, ir imprimiendo los números individualmente a medida que los vayas encontrando.

```python
import math
#Tengo mucho sueñoooo
primos = [2]

# Evaluamos cada número n desde 3 hasta 10000
for n in range(3, 10001):
    es_primo = True
    limite = math.sqrt(n)
    
    # Probamos divisibilidad solo con los primos encontrados hasta sqrt(n)
    for p in primos:
        if p > limite:
            break
        if n % p == 0:
            es_primo = False
            break  # Abandonamos las comprobaciones si encontramos un factor
            
    # Si no se encontró ningún factor, n es primo y lo agregamos a la lista
    if es_primo:
        primos.append(n)

# Imprimimos la lista completa al final o la cantidad de primos encontrados
print(f"Se encontraron {len(primos)} números primos hasta 10000.")
print(primos)
```

* **b)** Euclides demostró que el máximo común divisor $g(m, n)$ de dos enteros no negativos $m$ y $n$ satisface que:

  $$g(m, n) = \begin{cases} m & \text{si } n = 0, \\ g(n, m \bmod n) & \text{si } n > 0. \end{cases}$$

  Usando esta fórmula escribe una función `g(m, n)` que emplee **recursividad** para calcular el máximo común divisor de $m$ y $n$. Usa tu función para calcular e imprimir el máximo común divisor de 108 y 192.

```python
# INCISO A
def catalan(n):
    # Caso base
    if n == 0:
        return 1
    else:
        return ((4 * n - 2) * catalan(n - 1)) // (n + 1)

c_100 = catalan(100)
print(f"El valor de C_100 es:\n{c_100}")
```

```python
# INCISO B
def g(m, n):
    # Caso base: si n es 0, el MCD es m
    if n == 0:
        return m
    # Caso recursivo: g(n, m mod n)
    else:
        return g(n, m % n)

m = 108
n = 192

resultado = g(m, n)
print(f"El máximo común divisor de {m} y {n} es: {resultado}")
```
